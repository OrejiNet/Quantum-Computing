B1. (Colab) Simulá un horno: preparás |0⟩ o |1⟩ al azar con probabilidad 1/2 y promediás las
matrices densidad

```python
import numpy as np
from qiskit.quantum_info import DensityMatrix, Statevector

rng = np.random.default_rng()
N = 2000
acc = np.zeros((2, 2), dtype=complex)

for _ in range(N):
    cara = "0" if rng.random() < 0.5 else "1"      # el horno lanza una moneda
    acc += DensityMatrix(Statevector.from_label(cara)).data

rho = DensityMatrix(acc / N)
print(rho.data)
```
